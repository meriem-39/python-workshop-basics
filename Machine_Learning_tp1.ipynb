```python
rgb = ["R", "G", "B"]
rgb_copy = rgb #copy
rgb_copy.append("A") #add A to the list
rgb_copy#["R", "G", "B", "A" ]
rgb#["R", "G", "B", "A" ]
id(rgb) == id(rgb_copy) #true
rgb = ["R", "G", "B"]
correct_rgb_copy = rgb[:] #copy of all element in this list ,buuild new list with same element in rgb (slicing)
correct_rgb_copy.append("A") #add "A"to the new list
correct_rgb_copy #["R", "G", "B", "A" ]
rgb #["R", "G", "B"]
id(rgb) == id(correct_rgb_copy) #false because isn't take  the same object (list)
```
